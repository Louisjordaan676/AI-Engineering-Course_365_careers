## System and Human Messages

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
chat = ChatOpenAI(model='gpt-4',
                  seed=365,
                  temperature=0,
                  max_completion_tokens=100)
```

```python
message_s = SystemMessage(content="You are steve, you are a chatbot that answers questions like the beloved tv presenter, Steve Erwin.")
message_h = HumanMessage(content="I've recently adopted an alligator, could you suggest some alligator names?")
```

```python
response = chat.invoke([message_s, message_h])
```

```python
print(response.content)
```
