### Temperature of the model
- between 0 and 2
- the higher the more random
- defaults to 1

- models wit lower temps can be used when creating chatbots for educational purposes

```python
completion = client.chat.completions.create(model='gpt-4', messages= [
                                                                        {'role': 'user',
                                                                         'content': '''Could you explain briefly what a black hole is?'''}],
                                                                         max_tokens=250,
                                                                         temperature=0)
```

```python
print(completion.choices[0].message.content)
```

```python
completion = client.chat.completions.create(model='gpt-4', messages= [
                                                                        {'role': 'user',
                                                                         'content': '''Could you explain briefly what a black hole is?'''}],
                                                                         max_tokens=250,
                                                                         temperature=2)
```

```python
print(completion.choices[0].message.content)
# the model starts deviating very quickly... large temps = rarely helpful. what is the point of this then???
```

```python
completion = client.chat.completions.create(model='gpt-4', messages= [
                                                                        {'role': 'user',
                                                                         'content': '''Could you explain briefly what a black hole is?'''}],
                                                                         max_tokens=250,
                                                                         temperature=0,
                                                                         seed=365)
```

### Stream the answer that the LLM is generating


```python
completion = client.chat.completions.create(model='gpt-4', messages= [
                                                                        {'role': 'user',
                                                                         'content': '''Could you explain briefly what a black hole is?'''}],
                                                                         max_tokens=250,
                                                                         temperature=0,
                                                                         seed=365,
                                                                         stream=True) # as easy as setting stream to True
```

```python
completion
```

```python
for i in completion:
    print(i.choices[0].delta.content, end = "")
```
