re.sub()

searches for a specific text within a string and replaces it with new content

```python
string = r"sara was able to help me find the items I needed quickly"

# the name "sara" was spelled incorrectly. it was suppose to be "Sarah"
```

```python
# We can use re.sub() to replace it with the correct version

new_String = re.sub("sara", "sarah", string)
print(new_String)
```

So what can we use this for Louis?

suppose we have customer reviews, and we only want to see the ones about sarah, but we think some of the reviews might have spelled her name incorrectly


```python
customer_reviews = ['sam was a great help to me in the store', 
 'the cashier was very rude to me, I think her name was eleanor', 
 'amazing work from sadeen!', 
 'sarah was able to help me find the items i needed quickly', 
 'lucy is such a great addition to the team', 
 'great service from sara she found me what i wanted']
```

```python
# create an empty list to catch sarah's reviews
sarahs_reviews = []
```

```python
pattern_to_find = r"sarah?" # the question mark after the "h" means the "h is optional"
```

```python
for string in customer_reviews: #loops through each review
    if (re.search(pattern_to_find, string)):# basically checking for "sara" or "sarah" in the string
        sarahs_reviews.append(string) 
```

```python
print(sarahs_reviews)
```

Let's check for reviews that start with the letter "a"

```python
a_reviews = []
```

```python
pattern_to_find_a = r"^a" # the "^" symbol tells us to look at the start of the string
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find_a, string)):
        a_reviews.append(string)
```

```python
print(a_reviews)
```

find reviews that ends with the letter "y"

```python
y_reviews = []
pattern_to_find_y = r"y$"
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find_y, string)): #btw, this means that if this condition is True, it will preform what is said next
        y_reviews.append(string)
```

```python
print(y_reviews)
```

pipe symbol "|" works like an "or"

```python
needwant_reviews = []

pattern_to_find_nw = "r(need|want)ed"
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find_nw, string)): 
        needwant_reviews.append(string)
```

```python
print(needwant_reviews)
```

one of the most practical uses for regex is to remove punctuation from text.
this is crucial in NLP because it often gets in the way when we want to analyse words

```python
no_punct_reviews = []
pattern_to_find_punct = r"[^\w\s]" 
# the [] defines a set of characters we want to match
# "^" = "not"
# "\w" = word characters like underscores
# "\s" = whitespaces
# together this patter means "find anything that is not a word character and not a space"
```

```python
for string in customer_reviews:
    no_punct_string = re.sub(pattern_to_find_punct, "", string)
    no_punct_reviews.append(no_punct_string)
```

```python
print(no_punct_reviews)
```
