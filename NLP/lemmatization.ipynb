### LEMMATIZATION
uses a predefined dictionary to keep the context of the word

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer
```

```python
connect_tokens = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t, ": ", lemmatizer.lemmatize(t))
    #lemmatizer.lemmatize(t) takes the token t and looks it up in the 
    # wordnet dictionary to return its proper baseform
```

```python
for t in learn_tokens:
    print(f"{t}: {lemmatizer.lemmatize(t)}")
```

```python
for t in likes_tokens:
    print(f"{t}: {lemmatizer.lemmatize(t)}")
```
