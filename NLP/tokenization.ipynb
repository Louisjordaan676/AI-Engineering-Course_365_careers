Tokenization: breaking text into smaller units (tokens)

```python
import nltk
nltk.download('punkt_tab') # a table used to figure out where sentences begin and end
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sentences = "Her cat's name is Luna, Her dog's name is Max"
```

```python
sent_tokenize(sentences)
```

```python
sentece = "her cat's name is luna"
word_tokenize(sentece)
```

```python
sentence_2 = "Her cat's name is Luna and her dog's name is max"
word_tokenize(sentence_2)
```
